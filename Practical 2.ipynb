```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 40, 50, 78, 83])

m = 0
c = 0
learning_rate = 0.01

# Number of data points
n = len(x)

# Number of iterations
iterations = 100

# Gradient Descent
for i in range(iterations):

    y_pred = m * x + c

    dm = (-2 / n) * np.sum(x * (y - y_pred))
    dc = (-2 / n) * np.sum(y - y_pred)

    m_new = m - learning_rate * dm

    c = c - learning_rate * dc

    m = m_new

# Final predictions
y_pred = m * x + c

# Calculate Mean Square Error
mse = np.mean((y - y_pred) ** 2)

print("Final value of m =", m)
print("Final value of c =", c)
print("Mean Square Error =", mse)

# Plot actual data
plt.scatter(x, y, color="blue", s=60, label="Actual Data")

# Plot Gradient Descent line
plt.plot(x, y_pred, color="red", linewidth=2, label="Gradient Descent Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent with Mean Square Error")

plt.legend()
plt.grid()
plt.show()
```
