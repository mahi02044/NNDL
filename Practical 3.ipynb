```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Normalize data
X = (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0))

# Convert target into one-hot encoding
Y = np.zeros((len(y), 3))
Y[np.arange(len(y)), y] = 1

# Split data
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# Initialize weights and biases
np.random.seed(42)

w1 = np.random.rand(4, 5)
b1 = np.zeros(5)

w2 = np.random.rand(5, 3)
b2 = np.zeros(3)

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Derivative of sigmoid
def derivative_sigmoid(a):
    return a * (1 - a)

# Training settings
epochs = 5000
lr = 0.1

# Backpropagation
for epoch in range(epochs):

    for i in range(len(X_train)):

        x = X_train[i]
        target = Y_train[i]

        # Forward propagation
        h1 = (x @ w1) + b1
        a1 = sigmoid(h1)

        h2 = (a1 @ w2) + b2
        a2 = sigmoid(h2)

        # Calculate error
        loss = np.mean((target - a2) ** 2)

        # Backward propagation
        o2 = (a2 - target) * derivative_sigmoid(a2)
        o1 = derivative_sigmoid(a1) * (w2 @ o2)

        # Update weights and biases
        w2 -= lr * np.outer(a1, o2)
        b2 -= lr * o2

        w1 -= lr * np.outer(x, o1)
        b1 -= lr * o1

# Testing
correct = 0
predictions = []

for i in range(len(X_test)):

    x = X_test[i]

    # Forward propagation
    h1 = (x @ w1) + b1
    a1 = sigmoid(h1)

    h2 = (a1 @ w2) + b2
    a2 = sigmoid(h2)

    # Predicted class
    prediction = np.argmax(a2)
    predictions.append(prediction)

    actual = np.argmax(Y_test[i])

    if prediction == actual:
        correct += 1

# Accuracy
accuracy = correct / len(X_test) * 100

print("Actual Classes:")
print(np.argmax(Y_test, axis=1))

print("\nPredicted Classes:")
print(predictions)

print("\nAccuracy:", accuracy, "%")

# Final weights and biases
print("\nFinal W1:")
print(w1)

print("\nFinal B1:")
print(b1)

print("\nFinal W2:")
print(w2)

print("\nFinal B2:")
print(b2)
```
